# 1. 클래스가 필요한 이유

자동차 두 대를 변수와 함수만으로 관리해 봅시다.

````python
car1_brand = "현대"
car1_model = "아반떼"
car1_speed = 0

car2_brand = "기아"
car2_model = "K5"
car2_speed = 0


def accelerate(speed, amount):
    return speed + amount


car1_speed = accelerate(car1_speed, 30)
print(car1_model, car1_speed)
print(car2_model, car2_speed)
````

- 자동차가 100대면 변수가 300개 필요합니다.
- `car2_speed = accelerate(car1_speed, 30)` 같은 실수를 막을 방법이 없습니다.

**데이터(브랜드, 속도)와 기능(가속)이 흩어져 있는 게 문제입니다. 클래스는 이 둘을 하나로 묶어 줍니다.**

# 2. 클래스와 객체 기초

![자동차 클래스와 객체](https://velog.velcdn.com/images/newnew_daddy/post/d74d3483-fca1-49b9-9470-ee3c7256f9f0/image.png)

- **클래스**: 설계도. 어떤 **속성(데이터)** 과 **메서드(기능)** 를 가질지 정의합니다.
- **객체**: 설계도로 만든 실제 물건. 같은 설계도로 만들어도 **속성 값은 객체마다 다릅니다.**

빈 클래스에서 시작해 위 그림의 `Car` 를 한 단계씩 완성해 봅니다.

## 2-1. 빈 클래스와 객체 만들기

- `class 이름:` 으로 클래스를 정의합니다. (이름은 대문자로 시작)
- `이름()` 처럼 괄호를 붙여 호출하면 객체가 만들어집니다.

````python
class Car:
    pass


car1 = Car()
car2 = Car()

print(car1)
print(car2)
print(car1 == car2)
````

같은 설계도로 만들었지만 주소(`0x...`)가 다른 **서로 다른 객체**입니다.

````python
car3 = Car()
print(type(car3))   # <class '__main__.Car'>
````

## 2-2. 속성 붙이기

`객체.이름 = 값` 으로 객체에 데이터를 저장할 수 있습니다. 이것이 **속성**입니다.

````python
car1.brand = "현대"
car1.color = "흰색"

car2.brand = "기아"
car2.color = "검정"

print(car1.brand, car1.color)
print(car2.brand, car2.color)
````

**문제점:** 객체마다 속성을 하나하나 붙여야 하고, 빠뜨려도 모릅니다.

```python
car3 = Car()
car3.brand = "테슬라"    # color 를 깜빡했다
print(car3.color)        # AttributeError: 'Car' object has no attribute 'color'
```

→ **객체를 만들 때 속성을 한 번에 정하는 방법**이 필요합니다.

## 2-3. `__init__`: 만들 때 속성 정하기

- `__init__` 은 **객체가 만들어질 때 자동으로 실행**되는 메서드입니다. (앞뒤 밑줄 2개)
- `self` 는 **지금 만들어지고 있는 객체 자신**입니다.
- `self.brand = brand` → 받은 값을 객체의 속성으로 저장합니다.

````python
class Car:
    def __init__(self, brand, model, color):
        self.brand = brand
        self.model = model
        self.color = color


car1 = Car("현대", "아반떼", "흰색")
car2 = Car("기아", "K5", "검정")
car3 = Car("테슬라", "모델3", "빨강")

print(car1.brand, car1.model, car1.color)
print(car2.brand, car2.model, car2.color)
print(car3.brand, car3.model, car3.color)
````

`self` 자리에는 값을 넣지 않습니다. 파이썬이 자동으로 채워 줍니다.

````python
my_car = Car("BMW", "X5", "파랑")
print(my_car.brand, my_car.model, my_car.color)
````

## 2-4. 처음 값이 정해진 속성

모든 자동차는 속도 0에서 출발합니다. 이런 속성은 인자로 받지 않고 `__init__` 안에서 바로 정합니다.

````python
class Car:
    def __init__(self, brand, model, color):
        self.brand = brand
        self.model = model
        self.color = color
        self.speed = 0          # 인자로 받지 않는 속성


car1 = Car("현대", "아반떼", "흰색")
print(car1.model, car1.speed)
````

## 2-5. 메서드: 객체의 기능

- 클래스 안에 만든 함수를 **메서드**라고 합니다.
- 메서드의 첫 번째 자리는 항상 `self` 입니다.
- `self.속성` 으로 객체 자신의 데이터를 사용합니다.

````python
class Car:
    def __init__(self, brand, model, color):
        self.brand = brand
        self.model = model
        self.color = color
        self.speed = 0

    def info(self):
        print(f"{self.brand} {self.model} ({self.color}) - {self.speed}km/h")


car1 = Car("현대", "아반떼", "흰색")
car2 = Car("기아", "K5", "검정")

car1.info()
car2.info()
````

같은 `info()` 인데 결과가 다릅니다. `self` 가 각각 `car1`, `car2` 이기 때문입니다.

## 2-6. 속성을 바꾸는 메서드

메서드 안에서 `self.속성` 에 새 값을 넣으면 **그 객체의 속성만** 바뀝니다.

````python
class Car:
    def __init__(self, brand, model, color):
        self.brand = brand
        self.model = model
        self.color = color
        self.speed = 0

    def info(self):
        print(f"{self.brand} {self.model} ({self.color}) - {self.speed}km/h")

    def accelerate(self, amount):
        self.speed = self.speed + amount
        print(f"{self.model}: {amount}km/h 가속 → {self.speed}km/h")


car1 = Car("현대", "아반떼", "흰색")
car2 = Car("기아", "K5", "검정")

car1.accelerate(30)
car1.accelerate(20)
car2.info()             # car2 는 그대로 0km/h
````

## 2-7. `self` 의 정체

아래 두 줄은 **같은 동작**입니다. 점 앞의 객체가 `self` 로 들어갑니다.

```
car1.accelerate(10)
  │             │
 ▼             ▼
def accelerate(self, amount):
```

````python
car1.accelerate(10)
Car.accelerate(car1, 10)
````

## 2-8. `__str__`: `print()` 했을 때의 모양

객체를 그냥 `print()` 하면 `<__main__.Car object at 0x...>` 처럼 나옵니다.
`__str__` 이 돌려주는 문자열이 `print()` 결과가 됩니다.

````python
class Car:
    def __init__(self, brand, model, color):
        self.brand = brand
        self.model = model
        self.color = color
        self.speed = 0

    def info(self):
        print(f"{self.brand} {self.model} ({self.color}) - {self.speed}km/h")

    def accelerate(self, amount):
        self.speed = self.speed + amount
        print(f"{self.model}: {amount}km/h 가속 → {self.speed}km/h")

    def brake(self, amount):
        self.speed = self.speed - amount
        if self.speed < 0:
            self.speed = 0
        print(f"{self.model}: {amount}km/h 감속 → {self.speed}km/h")

    def __str__(self):
        return f"[{self.brand} {self.model}] {self.color} / {self.speed}km/h"


car1 = Car("현대", "아반떼", "흰색")
car2 = Car("기아", "K5", "검정")
car3 = Car("테슬라", "모델3", "빨강")

car3.accelerate(80)

print(car1)
print(car2)
print(car3)
````


## 2-9. 그 밖의 매직 메서드

`__init__`, `__str__` 처럼 **앞뒤로 밑줄 두 개**가 붙은 메서드를 **매직 메서드**라고 합니다.
직접 부르지 않아도 **파이썬이 특정 상황에서 자동으로 호출**합니다.

| 매직 메서드 | 자동으로 호출되는 때 | 예시 |
|---|---|---|
| `__init__` | 객체를 만들 때 | `Car("현대", ...)` |
| `__str__` | `print()`, `str()` | `print(car1)` |
| `__len__` | `len()` | `len(garage)` |
| `__getitem__` | 대괄호로 꺼낼 때 | `garage[0]` |
| `__eq__` | `==` 비교 | `car1 == car2` |
| `__lt__` | `<` 비교, `sorted()` | `car1 < car2` |
| `__add__` | `+` 연산 | `a + b` |

아래는 자동차를 담는 `Garage` 에 `__len__` 과 `__getitem__` 을 넣은 예시입니다.


````python

class Garage:
    def __init__(self):
        self.cars = []

    def add(self, car):
        self.cars.append(car)

    def __len__(self):              # len(garage)
        return len(self.cars)

    def __getitem__(self, index):   # garage[0]
        return self.cars[index]


garage = Garage()
garage.add(car1)
garage.add(car2)
garage.add(car3)

print(len(garage))   # 3
print(garage[0])     # car1 → Car 의 __str__ 로 출력
````

## 종합 실습. `Dog` 클래스

처음부터 직접 만들어 보세요.

- 속성: `breed`(견종), `age`(나이), `color`(색) → 만들 때 받음
- `info()`: `"말티즈 / 2살 / 흰색"` 형식으로 출력
- `birthday()`: 나이를 1 늘리고 `"말티즈: 3살이 되었습니다!"` 출력
- `bark()`: `"말티즈: 멍멍!"` 출력

````python
# 여기에 Dog 클래스를 작성하세요
class Dog:
    def __init__(self, breed, age, color):
        self.breed = breed
        self.age = age
        self.color = color

    def info(self):
        print(f"{self.breed} / {self.age}살 / {self.color}")

    def birthday(self):
        self.age += 1
        print(f"{self.breed}: {self.age}살이 되었습니다!")

    def bark(self):
        print(f"{self.breed}: 멍멍!")
````

````python
dog1 = Dog("나폴리탄 마스티프", 5, "검정")
dog2 = Dog("말티즈", 2, "흰색")
dog3 = Dog("차우차우", 3, "갈색")

dog1.info()
dog2.birthday()
dog3.bark()

# 기대 결과
# 나폴리탄 마스티프 / 5살 / 검정
# 말티즈: 3살이 되었습니다!
# 차우차우: 멍멍!
````

# 3. 클래스 상속

**부모 클래스의 속성과 메서드를 자식 클래스가 물려받는 것**입니다.
전기차는 자동차의 기능을 모두 가지면서, 배터리라는 기능이 더 있습니다.

- `class 자식(부모):` 로 상속받습니다.
- `super().__init__(...)`: 부모의 `__init__` 을 실행합니다.
- **오버라이딩**: 부모의 메서드를 자식이 같은 이름으로 다시 정의합니다.

````python
class Car:
    def __init__(self, brand, model, color):
        self.brand = brand
        self.model = model
        self.color = color
        self.speed = 0

    def info(self):
        print(f"{self.brand} {self.model} ({self.color}) - {self.speed}km/h")

    def accelerate(self, amount):
        self.speed = self.speed + amount
        print(f"{self.model}: {amount}km/h 가속 → {self.speed}km/h")

    def brake(self, amount):
        self.speed = self.speed - amount
        if self.speed < 0:
            self.speed = 0
        print(f"{self.model}: {amount}km/h 감속 → {self.speed}km/h")

    def __str__(self):
        return f"[{self.brand} {self.model}] {self.color} / {self.speed}km/h"
````

````python
class ElectricCar(Car):
    def __init__(self, brand, model, color, battery):
        super().__init__(brand, model, color)    # brand, model, color, speed 설정은 부모에게 맡김
        self.battery = battery                   # 전기차에만 있는 속성

    def charge(self):                            # 전기차에만 있는 메서드
        self.battery = 100
        print(f"{self.model}: 충전 완료 (배터리 {self.battery}%)")

    def info(self):                              # 오버라이딩
        print(f"{self.brand} {self.model} ({self.color}) - {self.speed}km/h, 배터리 {self.battery}%")


tesla = ElectricCar("테슬라", "모델3", "빨강", 40)

tesla.accelerate(50)   # Car 에게 물려받은 메서드
tesla.charge()         # ElectricCar 에만 있는 메서드
tesla.info()           # 다시 정의한 메서드
````
